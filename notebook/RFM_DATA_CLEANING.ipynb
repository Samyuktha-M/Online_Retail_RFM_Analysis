```python
#Importing dataset
import pandas as pd

df = pd.read_csv("/Users/samyukthamuralidharan/Desktop/Data Analysis projects for Portfolio/Customer Segmentation ( RFM analysis)/Online_Retail.csv",
                encoding='latin1')
```

```python
# Cleaning Online_Retail.csv 

# Change invoice_no to str
df['InvoiceNo']   = df['InvoiceNo'].astype(str)
```

```python
# Clean InvoiceDate
df['InvoiceDate'] = pd.to_datetime(df['InvoiceDate'])
```

```python
# Creating a separate data frame for cancellations
cancellations = df[df['InvoiceNo'].str.startswith('C')].copy()
df = df[~df['InvoiceNo'].str.startswith('C')]
```

```python
# Remove non-product StockCodes
product_mask = df['StockCode'].astype(str).str.match(r'^\d{5}[A-Za-z]?$')
df = df[product_mask]
```

```python
# Filter invalid quantity and price
df = df[df['Quantity']  > 0]
df = df[df['UnitPrice'] > 0]
```

```python
# Remove rows with missing CustomerID
df = df.dropna(subset=['CustomerID'])
```

```python
# Convert CustomerID to integer
df['CustomerID'] = df['CustomerID'].astype(int)
```

```python
# Drop exact duplicate rows
df = df.drop_duplicates()
```

```python
# Add Revenue column
df['Revenue'] = (df['Quantity'] * df['UnitPrice']).round(2)
```

```python
# Save cleaned file
df.to_csv("/Users/samyukthamuralidharan/Desktop/Data Analysis projects for Portfolio/Customer Segmentation ( RFM analysis)/Online_Retail_clean.csv", index=False)
```
